```python
# Importing numpy:
import numpy as np

# Importing tabulate:
from tabulate import tabulate
```

```python
# Set seed for reproductivity:
np.random.seed(42)
```

1. Data Simulation:


```python
# Creating a dummy dataset:
age = np.random.randint(20, 60, size = (5, 1))
salary = np.random.randint(30000, 150000, size = (5, 1))
house_price = np.random.randint(100000, 500000, size = (5, 1))


# combine columns horizonatally to make a 2D dataset:
dataset = np.hstack((age, salary, house_price))
table_dataset = tabulate(dataset, headers = ['Age', 'Salary', 'House_Price'], tablefmt = 'grid')

```

```python
print("--- RAW DATASET ---")
print(table_dataset)
print("-"*50)
```

2. Min-Max Normalization

```python
# Applying Min-Max normalization (0 and 1):
'''
Formula: X_new = (X - X_min) / (X_max - X_min)

where:
X_new = Min-Max normalization,
X_min = small element of X
X_max = big element of X
X = dataset
'''

def min_max_normalize(data):
  min_data = data.min(axis = 0)
  max_data = data.max(axis = 0)

  # Handle division by 0 if maximum = minimum
  range_data = np.where(max_data - min_data == 0, 1, max_data - min_data)
  return (data - min_data)/ range_data

normalized_dataset = min_max_normalize(dataset)
round_normalized_dataset = np.round(normalized_dataset, 4)
table_normalized_dataset = tabulate(round_normalized_dataset, headers = ['Age', 'Salary', 'Housing_Price'], tablefmt = 'grid')
```

```python
print("\n--- MIN-MAX NORMALIZED DATA ---")
print(table_normalized_dataset)
print("-"*50)
```

3. Z-Score Standaridzation (Mean = 0, Std Dev = 1)

```python
from os import scandir
# Apply Standardization(z-Score) fucntion:
'''
Formula: Z = X - X_mean / X_std

where:
Z = Standardization
X = data
X_mean = mean of data
X_std = std of data
'''

def z_score_standardization(data):
  mean_data = data.mean(axis = 0)
  std_data = data.std(axis = 0)

  # Handle division by 0 if std is zero
  std_data = np.where(std_data == 0, 1, std_data)
  return (data - mean_data) / std_data

standardized_dataset = z_score_standardization(dataset)
round_standardized_dataset = np.round(standardized_dataset, 4)
table_standardized_dataset = tabulate(round_standardized_dataset, headers = ['Age', 'Salary', 'House_Price'], tablefmt ='grid' )
```

```python
print("\n--- Z-SCORE STANDARDIZED DATA ---")
print(table_standardized_dataset)
print("-"*50)
```
